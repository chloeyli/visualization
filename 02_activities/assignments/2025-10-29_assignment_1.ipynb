```python
# making a line plot
# keep in mind, plot styles stay the same as before, unless I change it again

tipgraph = sns.lineplot(data = tips,
                        x='total_bill',
                        y='tip',
                        color='hotpink',
                        linestyle='-',
                        linewidth = 1,
                        marker = '*',
                        markerfacecolor = 'blue')

tipgraph.set(title = 'Tips vs. Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')

plt.show()
```

```python
# making a scatter plot
# seaborn isn't necessarily more complex than matplotlib, just can do more complex things with fewer lines of code

tipgraph = sns.scatterplot(data = tips, x = 'total_bill',
                           y='tip', style = 'time', 
                           hue = 'day', # 
                           palette = ['purple',
                            'hotpink', 'deepskyblue', 'yellowgreen'])

tipgraph.set(title = 'Tips vs. Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
import plotly.graph_objects as go # 'go' is ?
import nbformat

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
# make our figure
graph = go.Figure()

# choose our plot type (bar) and our data
graph.add_trace(go.Bar(x=x1, y=y1))
```

```python
# update our layour to include titles
graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score"
)

# show our plot
graph.show()
```

```python
# option to save file as html so we can embed onto sites
graph.write_html("../../folders/pirategraph.html")
```

```python
# make our figure
graph = go.Figure()

# choose our plot type (scatter) and our data
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', # we want points for a bar plot; a dot for each (x,y)
 marker=dict(
 size=15, # point size
 color='hotpink', # point colour
 opacity=1, # point transparency/alpha
 line=dict(width=5, color='purple') # point outline
 )))

graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score",
    width = 500,
    height = 500
)
```

```python
# making wordclouds
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv")
```

```python
text = " ".join(each for each in df.quote)

wordcloud = WordCloud(background_color="purple",
                      colormap = 'inferno',
                      random_state = 613).generate(text)

fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation = 'bilinear') # interpolation = squishing it into a better shape? if we are using chunkier data set...
ax.axis('off') # takes off the grid marks
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram = venn2([A,B],
                set_labels = ('Fruits', "Pies"),
                set_colors = ('blue', 'red'),
                alpha = 0.5)

# "10" --> binary for "show" on left circle; "not show" on right circle
# "\n" --> new line so banana, watermelon is in 2 lines
# "A-B" --> words only in A
diagram.get_label_by_id("10").set_text("\n".join(A-B)) ## left
diagram.get_label_by_id("11").set_text("\n".join(A & B)) ## middle slice
diagram.get_label_by_id("01").set_text("\n".join(B-A)) ## right
```

```python
diagram = venn2_unweighted([A, B],
 set_labels = ('Fruits', 'Pies'),
 set_colors=("blue", "red"),
 alpha=0.5)
```
