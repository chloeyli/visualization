```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.lineplot(data = tips,
             x= 'total_bill',
             y= 'tip')
```

```python
sns.set_style('dark')
```

```python
tipgraph = sns.lineplot (data=tips,
                         x='total_bill',
                         y='tip')
tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill($)',
             ylabel='Tip Amount ($)')

# what is the output: "[Text(0.5, 1.0, 'Tips vs. Total Bill'), Text(0.5, 0, 'Total Bill($)'),"?
# what's with the numbers Text(0.5, 1.0) or Text (0.5, 0)?
```

```python
fig = plt.subplots(figsize=(10,3))
# modified the size to be wider in x-axis (10) since above line plot looks pretty squished...
```

```python
# making a line plot

tipgraph = sns.lineplot(data = tips,
                        x='total_bill',
                        y='tip',
                        color='hotpink',
                        linestyle='-',
                        linewidth = 1,
                        marker = '*',
                        markerfacecolor = 'blue')
```

```python
# making a scatter plot

tipgraph = sns.scatterplot(data = tips, x = 'total_bill',
                           y='tip', style = 'time', 
                           hue = 'day', palette = ['purple', 
                            'hotpink', 'deepskyblue', 'yellowgreen'])

tipgraph.set(title = 'Tips vs. Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
sns.pairplot(
    data = tips,
    hue = 'day'
)
```

```python
daysplot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    kind="scatter",
    col_wrap=2)
```
