```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Markdown, display

sns.set_style("whitegrid")

# 1. load data set
# reading Toronto's Open Data on Mental Health Apprehensions from a csv file into a pandas dataframe
counts_df = pd.read_csv('../../02_activities/assignments/mental-health-apprehensions.csv')

# 2. exploring data set
counts_df.head()
counts_df.info()
```

```python
# 2. checking the dataset to see if there's null values
counts_df.isnull()
```

```python
# 2. check how many counts each category has:
counts_df['SEX'].value_counts()
```

```python
# 2. check how many counts each category has:
counts_df['NEIGHBOURHOOD_158'].value_counts()
```

```python
# 2. check how many counts each category has:
counts_df['OCC_DATE'].value_counts()
```

```python
# 2. check how many counts each category has:
counts_df['OCC_HOUR'].value_counts()
```

```python
# 3. preprocessing data: 
# turning OCC_DATE column from string to datetime
counts_df['OCC_DATE'] = pd.to_datetime(counts_df['OCC_DATE'])

# creating a new column Month with just the months from OCC_DATE
# code: hey pandas, treat this column as a bunch of dates, so i can access their month parts
counts_df['Month'] = counts_df['OCC_DATE'].dt.month

# seeing which months have highest apprehension rates
counts_df['Month'].value_counts()
```

```python
counts_df
```

```python
# 3. extracting season
counts_df['Season'] = counts_df['Month'].map({
    12:'Winter', 1:'Winter', 2:'Winter',
    3:'Spring', 4:'Spring', 5:'Spring',
    6:'Summer', 7:'Summer', 8:'Summer',
    9:'Autumn', 10:'Autumn', 11:'Autumn'
})

counts_df = counts_df[counts_df['SEX'] != 'Not Recorded']
# can't use .sum() here because the Season column is strings
season_counts = counts_df.groupby(['Season','SEX']).size().reset_index(name = 'Count')
```

```python
season_counts
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x='Season', 
            y='Count', 
            hue='SEX',
            data = season_counts,
            palette = 'dark')

plt.title("Seasons vs. Mental Health Act (MHA) Apprehension Counts",
          fontweight = 'bold',
          fontsize = 20)

plt.xlabel("Seasons",
           fontweight = 'bold',
           fontsize = 14)

plt.ylabel("MHA Apprehension Numbers",
           fontweight = 'bold',
           fontsize = 14)

plt.figtext(0.01, -0.05, 
         "Data Sourced from City of Toronto's Open Data Portal: (https://open.toronto.ca/dataset/mental-health-apprehensions/)",
         fontsize = 12,
         color = 'black')

plt.legend(loc = 'upper right')

plt.savefig("seasons_vs_counts.png", bbox_inches="tight")


```

```python
# define the order of categories i want seaborn to include (excluding 'not recorded')
# explicitly telling seaborn i only want female & male bars
gender = ['Female', 'Male']

# otherwise, figure showed 3 bars: female, male & not recorded (0)
sns.countplot(data = counts_df, 
              x = 'SEX', 
              order = gender,
              palette = ['Orange', 'Navy'])

plt.title("Sex vs. Mental Health Act (MHA) Apprehension Counts",
          fontweight="bold")
plt.xlabel("Sex of Apprehended Individuals",
           fontweight = 'bold')
plt.ylabel("MHA Apprehension Count Numbers",
           fontweight = 'bold')
plt.figtext(0.001, -0.05,
            "Data Sourced from City of Toronto's Open Data Portal (https://open.toronto.ca/dataset/mental-health-apprehensions/)")
```

# First Data Visualization: Using Python

```python
# reading csv file into a pandas dataframe
counts_df = pd.read_csv('../../02_activities/assignments/mental-health-apprehensions.csv')
counts_df.head()

# extracting just 2 columns of info here: Neighbourhood & Apprehension Count #'s
counts = counts_df['NEIGHBOURHOOD_158'].value_counts().reset_index() 
counts.columns = ['Neighbourhood', 'Apprehension Count']

# limiting dataframe to just top 20 neighbourhoods
top_20 = counts.head(20)

fig, ax = plt.subplots(figsize = (25,15))

top_20_graph = sns.barplot(data = top_20, 
                           x = 'Apprehension Count', 
                           y = 'Neighbourhood', color = 'navy', linewidth = 3)

# setting tick labels to be larger size
top_20_graph.tick_params(axis = 'x', 
                         labelsize = 25)
top_20_graph.tick_params(axis = 'y', 
                         labelsize = 25)

fig.suptitle('Top 20 Toronto Neighbourhoods by Mental Health Act (MHA) Apprehension Counts', fontsize = 36, fontweight = 'bold', x=0.5, y=0.95)
top_20_graph.set_xlabel('MHA Apprehension Numbers', fontsize = 30, fontweight = 'bold')
top_20_graph.set_ylabel('Toronto Neighbourhoods', fontsize = 30, fontweight = 'bold')

# leave space for title
# rect=[0,0,1,0.95] tells Matplotlib:
# lay out the axes so they fill the figure from left to right (0 → 1) and bottom to 95% of the height (0 → 0.95)
# The top 5% of the figure is left free for the title
plt.tight_layout(rect=[0, 0, 1, 0.95])  

# fig.text(x, y, text, ha='left', va='bottom')
# x is in fig coordinates, range from 0 (left) → 1 (right) of page
#  - 'right': means right edge of text is at x 
fig.text(0.99, -0.05, 
         "Data Sourced from City of Toronto's Open Data Portal: (https://open.toronto.ca/dataset/mental-health-apprehensions/)",
         fontsize = 24,
         color = 'black',
         ha = 'right',
         va = 'bottom')

# want to do an alt text for this figure, so I'm saving it first as apprehension_counts.png
plt.savefig("apprehension_vs_counts.png", bbox_inches="tight")

# showing the plot
plt.show()
```
